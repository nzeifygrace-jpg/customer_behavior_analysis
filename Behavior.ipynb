```python
#Checking for null values

df.isnull().sum()
```

```python
#To address the null valus at Review Rating,I will use the Medium value of each category.

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#To rewrite all column names to a consistent lowercase and replace the spaces with 
#snake style (_) for ease of analysis and error reduction...

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#creating new column called age_group

labels = ['Young Adults', 'Adults', 'Middle Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
#creating another column called frequency_of_purchases_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 months': 90
}

df['frequency_of_purchases_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
